# SIH26142 — Data Exploration
Quick visual checks on LR/HR pairs before committing to a training run.

```python
import sys
sys.path.append('../src')
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
# Load a sample LR/HR pair and eyeball it before training
lr_paths = sorted(Path('../data/processed/lr').glob('*.npy'))
hr_paths = sorted(Path('../data/processed/hr').glob('*.npy'))
print(f"{len(lr_paths)} LR patches, {len(hr_paths)} HR patches")
```

```python
if lr_paths:
    lr = np.load(lr_paths[0])
    hr = np.load(hr_paths[0])
    fig, ax = plt.subplots(1, 2, figsize=(10, 5))
    ax[0].imshow(lr[..., :3]); ax[0].set_title(f'LR {lr.shape}')
    ax[1].imshow(hr[..., :3]); ax[1].set_title(f'HR {hr.shape}')
    plt.show()
else:
    print('No patches yet — run src/data_pipeline.py first.')
```
